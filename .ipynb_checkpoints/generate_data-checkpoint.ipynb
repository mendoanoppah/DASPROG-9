```python
import numpy as np
import pandas as pd

np.random.seed(42)

rows = 10000

data = {
    'ID': np.arange(1, rows+1),
    'nama': np.random.choice(['sopo', 'rixki', 'asep', 'ujang', 'wiliam'], rows),
    'usia': np.random.randint(18, 25, rows),
    'jenis_kelamin': np.random.choice(['laki-laki', 'perempuan'], rows),
    'pekerjaan': np.random.choice(['cyber security', 'devops', 'data scientist', 'cloud computing', 'software engineer'], rows),
    'gaji': np.random.normal(100000000, 10000000, rows).astype(int),
    'status_menikah': np.random.choice(['menikah', 'belum menikah'], rows),
    'jumlah_anak': np.random.randint(0, 3, rows),
    'tahun_pengalaman': np.random.randint(0, 10, rows),
    'lokasi': np.random.choice(['bogor', 'bandung', 'malang', 'sukabumi'], rows)
}

df = pd.DataFrame(data)
df.head()

# print(df.info())
# print(df.describe(include="all"))
```
